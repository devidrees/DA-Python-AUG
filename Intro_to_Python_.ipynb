- Sequence Data Types
- this is data which has a sequence, one followed by another in a well defined order

````python
# Loops


for x in range(0,10,2): # range(start, stop, stepsize)
  print(x)
````

````python
for ch in "idrees":
  print(ch)
````

````python
for id, val in enumerate("idrees"):
  print(id, val)
  if val == "r":
    break
````

````python
# while

count_flag = 1

while True:
  # you need the stop condition
  if count_flag == 10:
    break

  print("idrees ", count_flag)
  count_flag = count_flag + 1 # count_flag += 1
````

- Lists store **POINTERS**(addresses)

- Structures that: **ARRAY**
  - saves actual data?
  - Fast
  - but it will not be flexible

````python
[2,3,4,5,"idrees", [1,2,3,4], {1,2,3}]
# address is saved for everything
# real location saves data
````
