## Lambda Fuction
- small, one line fuctions, anonymous function

```python
lambda parameters : expression

variable_name = lambda x, y : <expression of x and y>
```


````python
# Lambda Functions
# to calculate cube of a number

#   parameter
#     input        return
cube = lambda  x:   x**3 # exponent **

#  argument - real input
cube(3)


### give me a lambda function to calculate addition of two numbers
add = lambda  x,y:   x+y

add(2,3) # result in 5

## lambda function to double a number
double = lambda  x:   x*2
double(100)
````
